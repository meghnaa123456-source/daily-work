```python
import matplotlib.pyplot as plt

months = ["jan","feb","mar","apr","may","jun"]
electronics = [120,150,180,170,210,240]
clothing = [80,90,100,120,130,150]
furniture = [50,60,70,65,80,90]

plt.stackplot(
    months,
    electronics,
    clothing,
    furniture,
    labels=["electronics","clothing","furniture"]
)

plt.xlabel("month")
plt.ylabel("sale(₹ thousands)")
plt.title("monthly sales by product category")

plt.legend()
plt.show()
```

```python
plt.plot(clothing,furniture)
plt.text(120,80,"important point")
plt.show()
```
