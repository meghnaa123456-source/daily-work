```python

plt.figure(figsize=(18,10))

# 1. Bar chart
x = df.groupby('Category')['Quantity'].mean()
plt.subplot(2,3,1)
plt.bar(x.index, x.values)
plt.title("Category vs Quantity")
plt.xlabel("Category")
plt.ylabel("Quantity")

# 2. Boxplot
plt.subplot(2,3,2)
plt.boxplot(df['Unit_Price'].dropna())
plt.title("Unit Price")
plt.ylabel("Price")

# 3. Countplot
plt.subplot(2,3,3)
sns.countplot(x='Product', data=df)
plt.title("Countplot")
plt.xlabel("Product")
plt.ylabel("Count")
plt.xticks(rotation=45)

# 4. Histogram
plt.subplot(2,3,4)
plt.hist(df['Quantity'], bins=5)
plt.title("Quantity Distribution")
plt.xlabel("Quantity")
plt.ylabel("Frequency")

# 5. Scatterplot
plt.subplot(2,3,5)
plt.scatter(df['Quantity'], df['Unit_Price'])
plt.title("Quantity vs Unit Price")
plt.xlabel("Quantity")
plt.ylabel("Unit Price")

# 6. Line plot
plt.subplot(2,3,6)
x = df.groupby('Quantity')['Discount'].sum()
plt.plot(x.index, x.values)
plt.title("Quantity vs Discount")
plt.xlabel("Quantity")
plt.ylabel("Discount")

plt.tight_layout()
plt.show()
```

```python
plt.Figure(figsize=(8,5))
sns.heatmap(
    df.corr(numeric_only=True),annot=True,cmap="coolwarm"
)
plt.title("correlation heatmap")
plt.show()
```
