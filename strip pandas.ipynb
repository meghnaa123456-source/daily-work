```python
import pandas as pd

data={
    "Customer":["Amit","PRIYA","rahul","Amit"],
    "Email":["amit@gmail.com","priya@company.org","rahul123@yahoo.com","sneha@GMAIL.com"],
    "Phone":["+91-9876543210","987643211","91-9876543212","9876543213"],
    "Amount":["$1,200","1500 INR","Rs.2,000","1500"]
}
df=pd.DataFrame(data)
df
```

```python

df["Customer"]=df["Customer"].str.strip()
df
```

```python
df["Customer"]=df["Customer"].str.lower()
df
```

```python
df["Amount"]=df["Amount"].str.replace(",","")
df
```

```python
df["Amount"]=df["Amount"].str.replace("Rs.","")
df
```

```python
df["Amount"]=df["Amount"].str.replace(r"[$INR .-]","",regex=True)
df
```

```python
df["Phone"]=df["Phone"].str.replace("91","")
df
```

```python
df["Phone"]=df["Phone"].str.replace("+","")
df
```

```python
df["Phone"]=df["Phone"].str.replace("-","")
df
```
